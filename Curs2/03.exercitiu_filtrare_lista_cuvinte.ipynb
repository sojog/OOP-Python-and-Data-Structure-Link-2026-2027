# Varianta cu list comprehention

```python
def filtrare_cuvinte_scurte(o_lista:list[str]) -> list[str]:
    noua_lista = [cuv for cuv in o_lista if len(cuv) > 5]
    return noua_lista


filtrare_cuvinte_scurte(["rau", "bun", "ploaie", "caine", "mancare", "apa", "hartie"])
```

# Parametru cu valoare default

```python
def filtrare_cuvinte_scurte(o_lista:list[str], minim_caractere) -> list[str]:
    noua_lista = []
    for cuv in o_lista:
        if len(cuv) > minim_caractere:
            noua_lista.append(cuv)

    return noua_lista


filtrare_cuvinte_scurte(["rau", "bun", "ploaie", "caine", "mancare", "apa", "hartie"], minim_caractere=5)
```

# Definirea unui parametru default

```python
def filtrare_cuvinte_scurte(o_lista:list[str], minim_caractere=5) -> list[str]:
    noua_lista = []
    for cuv in o_lista:
        if len(cuv) > minim_caractere:
            noua_lista.append(cuv)

    return noua_lista


filtrare_cuvinte_scurte(["rau", "bun", "ploaie", "caine", "mancare", "apa", "hartie"], minim_caractere=5)
```
