# args vs kargs

```python
def functie(x,y,*args):
    print("x=", x)
    print("y=", y)
    print("args=", args, type(args))

functie(1, 2, 3, 4)
```

```python
def functie(*args):
    print("args=", args, type(args))

functie(1, 2, 3)
```

```python
def functie(*args, **kargs):
    print("args=", args, type(args))
    print("kargs=", kargs, type(kargs))

functie(1, 2, 3)
```

```python
functie(alb=1, negru=2, verde=3)
```

```python
functie(1, 2, 3, 4, alb=1, negru=2, verde=3)
```

# Intotdeauna args apoi kargs
