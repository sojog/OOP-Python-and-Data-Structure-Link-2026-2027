```python
class Produs:
    def __init__(self, nume, pret):
        self.nume = nume
        self.pret = pret

    # o functie a clasei se numeste metoda, iar primul parametru este self
    def pret_cu_tva(self):
        return self.pret * 1.21


obiect = Produs("oglinda", 100)
obiect.pret_cu_tva()
```

```python
class Produs:
    def __init__(self, nume, pret):
        self.nume = nume
        self.pret = pret

    # o functie a clasei se numeste metoda, iar primul parametru este self
    def pret_cu_tva(self):
        return self.pret * 1.21


    def reducere_10_procente(self):
        self.pret -= 0.1 * self.pret


oglinda = Produs("Oglinda de perete", 100)
print(oglinda.pret_cu_tva())


telefon = Produs("iPhone", 2000)
print(telefon.pret_cu_tva())
```
