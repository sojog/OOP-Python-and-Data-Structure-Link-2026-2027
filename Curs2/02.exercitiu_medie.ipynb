```python
def media_cu_lista(o_lista:list):
    print("lista:", o_lista)
    suma = 0
    nr_elemente = 0
    for i in o_lista:
         nr_elemente += 1
         suma += i

    if nr_elemente > 0:
         return suma / nr_elemente
  # in cazul acesta exista pericolul impartirii la zero
    return 0

def media_cu_args(*args):
    print("args:", args)
    suma = 0
    nr_elemente = 0
    for i in args:
            nr_elemente += 1
            suma += i

    if nr_elemente > 0:
            return suma / nr_elemente # in cazul acesta exista pericolul impartirii la zero
    return 0
```

```python
media_cu_lista([1, 2, 3])
```

```python
media_cu_args(1, 2, 3)
```
