```python
def al_doilea_cel_mai_mic_din(o_lista:list[int]) -> int:
    lista_sortata = sorted(o_lista)
    if len(lista_sortata) > 1:
        return lista_sortata[1]
    elif lista_sortata:
        return lista_sortata[0]
    return "Lista goala"
```

```python
al_doilea_cel_mai_mic_din([10, 2, 4, 5])
```
