```python
class Account:
    def __init__(self, suma_initiala=0):
        self.sold = suma_initiala

    def withdraw(self, suma):
        if suma > self.sold:
            print("Fonduri insuficiente")
        else:
            self.sold -= suma

cont1 = Account(suma_initiala=100) # suprascierea parametrului default suma_initiala=0
print(cont1.sold)

cont1.withdraw(10)
print(cont1.sold)

cont1.withdraw(10)
print(cont1.sold)

cont1.withdraw(100)
print(cont1.sold)
```

```python
class Account:
    def __init__(self, suma_initiala=0):
        self.sold = suma_initiala

    def withdraw(self, suma):
        if suma > self.sold:
            print("Fonduri insuficiente")
        elif suma < 0:
            return
        else:
            self.sold -= suma

    def add_money(self, suma):
        if suma < 0: return
        self.sold += suma

cont1 = Account(suma_initiala=100) # suprascierea parametrului default suma_initiala=0
print(cont1.sold)

cont1.withdraw(10)
print(cont1.sold)

cont1.withdraw(10)
print(cont1.sold)

cont1.withdraw(100)
print(cont1.sold)
```

```python
class Account:
    def __init__(self, suma_initiala=0):
        self.sold = suma_initiala

    def withdraw(self, suma):
        if suma > self.sold:
            print("Fonduri insuficiente")
        elif suma < 0:
            return
        else:
            self.sold -= suma

    def add_money(self, suma):
        if suma < 0: return
        self.sold += suma

    def show_balance(self):
        print("Suma disponibila este:", self.sold)

cont1 = Account(suma_initiala=100) # suprascierea parametrului default suma_initiala=0
cont1.show_balance()

cont1.withdraw(80)
cont1.show_balance()

cont1.add_money(100)
cont1.show_balance()
```
