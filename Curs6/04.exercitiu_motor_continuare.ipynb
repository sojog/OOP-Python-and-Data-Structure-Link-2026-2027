```python
class Motor:
    def __init__(self, seria, putere, nr_km):
        self.__seria = seria
        self.__putere = putere
        self.__km = nr_km

    def __str__(self):
        return f"Seria {self.seria} cu puterea {self.putere} si {self.km} km"

    def __repr__(self):
        return self.__str__()


    def __int__(self):
        return self.km

    # obiect += 10
    # obiect = obiect + 10
    def __iadd__(self, distanta):
        if isinstance(distanta, int) and distanta > 0:
            self.km += distanta
        return self

    @property
    def seria(self):
        return self.__seria

    @property
    def putere(self):
        return self.__putere

    @property
    def km(self):
        return self.__km

    @km.setter
    def km(self, new_value):
        if isinstance(new_value, int) and new_value > self.km:
            self.__km = new_value
```
