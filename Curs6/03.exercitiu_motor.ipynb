```python
class Motor:
    def __init__(self, seria, putere, nr_km):
        self.__seria = seria
        self.__putere = putere
        self.__km = nr_km

    @property
    def seria(self):
        return self.__seria

    @property
    def putere(self):
        return self.__putere

    @property
    def km(self):
        return self.__km

    @km.setter
    def km(self, new_value):
        if isinstance(new_value, int) and new_value > self.km:
            self.__km = new_value
```

```python
motor1 = Motor("abc", 100, 50_000)
motor1
```
