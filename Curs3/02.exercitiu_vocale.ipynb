```python
text = "Salutare, ce mai faci?"

nr_vocale = 0
nr_consoane = 0

for ch in text:
    if ch in vocale:
        nr_vocale += 1
    elif ch.isalpha():
        nr_consoane += 1

print("nr_vocale", nr_vocale)
print("nr_consoane", nr_consoane)
```

# Transformare in functie

```python
def analiza_text (text):
    litere = {}
    vocale = 'aeiouAEIOU'
    for ch in text:
        if ch in vocale:
            litere["vocale"] = litere.get("vocale", 0) +  1
        elif ch.isalpha():
            litere["consoane"] = litere.get("consoane", 0) + 1
    return litere
```

```python
analiza_text ("Salutare, ce mai faci?")
```

```python
analiza_text ("Buna seara, ce mai faci?")
```
