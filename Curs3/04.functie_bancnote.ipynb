```python
def calculeaza_bancnote(suma):
    bancnote = [10, 5, 2, 1]

    rezultat = {}

    for banc_val in bancnote:
        nr_bancnote = suma // banc_val
        print(f"Nr bancnote de {banc_val} este {nr_bancnote}")
        rezultat[banc_val] = nr_bancnote
        suma = suma % banc_val

    return rezultat

calculeaza_bancnote(128)
    
```
