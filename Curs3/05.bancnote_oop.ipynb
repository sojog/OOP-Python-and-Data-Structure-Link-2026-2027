```python
class NumaratorBani:
    def __init__(self, moneda, tipuri_bancnote):
        self.moneda = moneda
        ## Curatarea inputului
        self.bancnote = sorted(tipuri_bancnote, reverse=True)

    def calculeaza_bancnote(self, suma):
        rezultat = {}
        for banc_val in self.bancnote:
            nr_bancnote = suma // banc_val
            # print(f"Nr bancnote de {banc_val} este {nr_bancnote}")
            rezultat[banc_val] = nr_bancnote
            suma = suma % banc_val

        return rezultat
```

```python
numara_ron = NumaratorBani("RON", [100, 500, 200,  50, 20, 10, 5, 1])
numara_ron.calculeaza_bancnote(789)
```
