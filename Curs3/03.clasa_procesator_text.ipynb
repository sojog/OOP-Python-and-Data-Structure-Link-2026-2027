```python
class ProcesatorText: # CamelCase
    
    #  Ce are (atribute/variabile ale clasei) - textul efectiv, vocalele
    def __init__(self, text_initial):
        self.text = text_initial
        self.vocale = "aeiou" + "aeiou".upper()

    # Ce face ( metode/functii ale clasei) - fara consoane, fara vocale, inversare
    def text_fara_consoane(self):
        nou_text = ""
        for ch in self.text:
            if ch.isalpha() and ch in self.vocale:
                nou_text += ch
        return nou_text

    def text_fara_vocale(self):
        nou_text = ""
        for ch in self.text:
            if ch not in self.vocale:
                nou_text += ch
        return nou_text

    def inverseaza_text(self):
        return self.text[::-1]
```
