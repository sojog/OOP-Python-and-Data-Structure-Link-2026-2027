# Variabile locale vs variabile globale

```python
# Global
x = 1

def o_functie():
    # Local
    x = 100
    print("Functia a fost apelata")
```

```python
o_functie()
```

```python
x
```

```python
x
```

```python
def o_alta_functie():
    # Local
    un_alt_x = 100
    print("Functia a fost apelata")
```

```python
o_alta_functie()
```

```python
un_alt_x
```

```python
def inca_o_alta_functie():
    # Global - este un keyword, face ca o variabila global sa fie schimbata intr-o functie
    global inca_un_x
    inca_un_x = 100
    print("Functia a fost apelata")
```

```python
inca_un_x
```

```python
inca_o_alta_functie()
```
