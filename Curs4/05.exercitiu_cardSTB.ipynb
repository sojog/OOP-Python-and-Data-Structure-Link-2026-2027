```python
class CardSTB:
    # valoare statica (apartine de clasa)
    COST_CALATORIE = 3

    def __init__(self, nume = "Nenominal", calatorii = 0, credit = 0):
        self.__nume = nume
        self.__calatorii = calatorii
        self.__credit = credit

    def __str__(self):
        return f"{self.__nume} are {self.__calatorii} calatorii si {self.__credit} RON credit"

    def __repr__(self):
        return f"{self.__nume} are {self.__calatorii} calatorii si {self.__credit} RON credit"

    def __iadd__(self, suma):
        if isinstance(suma, int) and suma > 0:
            self.__credit += suma 
        return self

    def alimentare_calatorii(self, nr_calatorii = 10):
        if isinstance(nr_calatorii, int) and nr_calatorii >= 10:
            self.__calatorii += nr_calatorii



    def validare(self):
        if self.__calatorii > 0:
            self.__calatorii -= 1
        elif self.__credit > CardSTB.COST_CALATORIE:
            self.__credit -= CardSTB.COST_CALATORIE
        else:
            print("Credit insuficient")

card1 = CardSTB(credit=10)
card1
```
