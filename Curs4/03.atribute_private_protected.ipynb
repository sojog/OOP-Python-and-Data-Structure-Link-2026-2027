```python
class Point:
    def __init__(self, x:int, y:int):
        self.x = x
        self.y = y
        
    def __str__(self):
        return f"({self.x}, {self.y})"

    def __bool__(self):
        return (self.x != 0) or (self.y != 0)

    def set_to_origin(self):
        self.x = 0
        self.y = 0

    def __iadd__(self, other):
        if isinstance(other, Point):
            self.x += other.x
            self.y += other.y
        elif isinstance(other, tuple):
            dx, dy = other
            self.x += dx
            self.y += dy
        return self
```

```python
pct1 = Point(2, 3)
print(pct1)
```

# Declararea unor atribute protected/private 

## Atribut protected/internal -> incepe cu un "_" (Nu trebuie schimbat in afara clasei, desi este posibil technic vorbind)

```python
class Point:
    def __init__(self, x:int, y:int):
        self._x = x
        self._y = y
        
    def __str__(self):
        return f"({self._x}, {self._y})"

    def __bool__(self):
        return (self._x != 0) or (self._y != 0)

    def set_to_origin(self):
        self._x = 0
        self._y = 0

    def __iadd__(self, other):
        if isinstance(other, Point):
            self._x += other._x
            self._y += other._y
        elif isinstance(other, tuple):
            dx, dy = other
            self._x += dx
            self._y += dy
        return self
```

```python
pct1 = Point(2, 3)
print(pct1)
```

```python
pct1._x
```

```python
pct1._x = "orice"
```

```python
pct1 += (10, 10)
```
