```python
def este_palindrom(cuvant):
    for i in range(len(cuvant) // 2):
        if cuvant[i] != cuvant[-i-1]:
            print("Cuvantul nu este palindrom")
            break
    else: # Acest else se executa doar cand for-ul ajunge la final (nu exista un BREAK)
        print("Cuvantul este palindrom")


este_palindrom("caiac")
```

```python
def este_palindrom(cuvant):
    for i in range(len(cuvant) // 2):
        if cuvant[i] != cuvant[-i-1]:
            return f"Cuvantul {cuvant} nu este palindrom"
            
    else: # Acest else se executa doar cand for-ul ajunge la final (nu exista un BREAK)
        return f"Cuvantul {cuvant} nu este palindrom"


este_palindrom("caiac")
```
