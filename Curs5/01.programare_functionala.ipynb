# In Python functiile pot fi pasate ca si parametri

```python
def adunare(a, b):
    return a + b
```

```python
adunare
```

```python
suma = adunare
```

```python
suma(2, 3)
```

```python
def scadere(a, b):
    return a - b

def inmultire(a, b):
    return a * b
```

```python
def exercuta_operatia(functia, a, b):
    return functia(a, b)
```

```python
exercuta_operatia(inmultire, 6, 7)
```

```python
exercuta_operatia(scadere, 6, 7)
```

```python
o_lista = ["Amsterdam", "Bucuresti", "Iasi", "Constanta", "Arad", "Bacau"]

sorted(o_lista)
```

```python
for i in o_lista:
    print("i=", i, "len(i)", len(i))
```

```python
sorted(o_lista, key = len )
```

# Functii anonime - functii folosite o singura data

```python
def adunare(a, b):
    return a + b
```

Functia poate fi pasata ca parametru (similar cu o variabila)

```python
o_alta_functie_adunare = adunare
```

```python
o_alta_functie_adunare(2, 10)
```

```python
arata_in_consola = print
```

```python
arata_in_consola("Hello World")
```

# O alta metoda de a defini o functie 

```python
aduna_doi = lambda x : x + 2
aduna_doi
```

```python
aduna_doi(10)
```

```python
def aduna_doi(x):
    return x + 2
```

```python
lambda x, y : x ** y
```

```python
putere = lambda x, y : x ** y
```

```python
putere(2, 3)
```

```python
o_lista = ["Amsterdam", "Bucuresti", "Iasi", "Constanta", "Arad", "Bacau"]
```

```python
sorted(o_lista, key=len)
```

```python
sorted(o_lista, key=lambda x:x[1])
```
