```python
class TimeConverter: # CamelCase
    def __init__(self, hours, minutes):
        self.__hours = hours
        self.__minutes = minutes

    def __str__(self):
        return f"{self.__hours} ora si {self.__minutes} minute"

    def __repr__(self):
        return self.__str__()

    def to_minutes(self):
        return self.__hours * 60 + self.__minutes

    def to_hours(self):
        return self.__hours + self.__minutes / 60

    def add_time(self, other):
        return self.to_minutes() + other.to_minutes()

    def diff_time(self, other):
        return self.to_minutes() - other.to_minutes()
```

```python
t4 = TimeConverter(1, 2)
t5 = TimeConverter(2, 6)
```

```python
t4.diff_time(t5)
```

```python
t5.diff_time(t4)
```

```python
t5 - t4 
```

```python
t5 + t4 
```
